```python
from scipy import stats

n        = 200
k        = 9
p_chance = 5 / 200   # 0.025

# One-sided binomial test (for the p-value)
result  = stats.binomtest(k, n, p_chance, alternative='greater')
p_value = result.pvalue

# Two-sided Wilson CI (for the interval)
ci = stats.binomtest(k, n).proportion_ci(
        confidence_level=0.95, method='wilson')

print(f"Exact p-value (one-sided): {p_value:.4f}")
print(f"Expected correct at chance: {n * p_chance:.1f}")
print(f"Observed correct: {k}")
print(f"Ratio: {k / (n * p_chance):.2f}x")
print(f"\n95% Wilson CI (two-sided): [{ci.low:.4f}, {ci.high:.4f}]")
print(f"Chance rate: {p_chance:.4f}")

if ci.low > p_chance:
    print("-> Chance is BELOW the CI lower bound (result IS significant)")
else:
    print("-> Chance is WITHIN the CI (result is NOT significant)")
```
